```python
import pandas as pd
import re

dicionario = {'fruta': ['maçã', 'banana', 'laranja'],
              'animal': ['cachorro', 'leão', 'peixe'],
              'veiculo': ['carro', 'moto', 'onibus']}

texto = "meu cachorro gosta de passear de carro e de comer maçã."
padrao_sem_pontuacao = r'[^\w\s]'
texto = re.sub(padrao_sem_pontuacao, '', texto)

contagem_categorias = {key: 0 for key in dicionario.keys()}
palavras = texto.lower().split()

# Continuar a partir desse trecho.
# Iterate over each word in the text
for palavra_texto in palavras:
    # Iterate over each category and its associated words in the dictionary
    for categoria, palavras_associadas in dicionario.items():
        # Check if the word from the text is in the current category's list of words
        if palavra_texto in palavras_associadas:
            contagem_categorias[categoria] += 1

# Print the results
print("Contagem de palavras por categoria:")
for categoria, count in contagem_categorias.items():
    print(f"  {categoria.capitalize()}: {count}")
```

```python
import re

texto = "No dia 07/07/1822, foi proclamada a independência do Brasil e no dia 15/11/1899, foi proclamada a república no Brasil"

padrao = r'\b(\d{1,2})/(\d{1,2})\b'

combinacoes = re.findall(padrao, texto)

datas = []
for d in combinacoes:
  dia = int(d[0])
  mes = int(d[1])
  datas.append((dia, mes))

  print(datas)
```

```python

import re

palavras_positivas = ['bom', 'excelente', 'feliz']
palavras_negativas = ['ruim', 'péssimo', 'triste']

texto = "O filme foi excelente. De fato, foi muito bom, mas teve um final triste."

qtd_positiva = len(re.findall(r'\b(?:%s)\b' % '|'.join(palavras_positivas), texto, flags=re.IGNORECASE))
qtd_negativa = len(re.findall(r'\b(?:%s)\b' % '|'.join(palavras_negativas), texto, flags=re.IGNORECASE))

sentimento='Neutro'
if qtd_positiva > qtd_negativa:
  sentimento='Positivo'
elif qtd_negativa > qtd_positiva:
   sentimento='Negativo'

print(f'O sentimento sobre o filme foi: {sentimento}')
```

### Removing specific words and short words

Building upon the previous examples, this code constructs a single regular expression to remove both a list of specified words and any words that are two characters or less in length. It then cleans up any extra spaces left after the replacements.

```python
# Define the initial text and the specific words to eliminate
texto_original = "Este texto é um exemplo para fazer um checklist de revisão do motor do carro."
palavras_para_eliminar = ["este", "para", "de"]

# Convert the text to lowercase to ensure case-insensitive matching
texto_lower = texto_original.lower()

# Prepare the specific words for the regex pattern by escaping them and joining with '|'
specific_words_pattern_part = '|'.join(re.escape(p) for p in palavras_para_eliminar)

# Create a combined regex pattern:
# 1. Matches any of the specific words (e.g., 'este', 'para', 'de')
# 2. OR matches any word composed of 1 or 2 word characters (e.g., 'é', 'um', 'do', 'o')
# \b ensures full word matching
combined_pattern = rf'\b({specific_words_pattern_part}|\w{{1,2}})\b'

# Apply the regex substitution, replacing matched words with a single space
texto_com_espacos = re.sub(combined_pattern, ' ', texto_lower)

# Clean up any multiple spaces resulting from the replacements and strip leading/trailing spaces
texto_final = re.sub(r'\s+', ' ', texto_com_espacos).strip()

print(f"Texto original: '{texto_original}'")
print(f"Palavras a serem eliminadas: {palavras_para_eliminar} e palavras com 1 ou 2 caracteres.")
print(f"Texto processado: '{texto_final}'")
```
